# Model Evaluation
Compare baseline vs. new algorithm on the validation set.

```python
# training cell
import matplotlib.pyplot as plt

baseline_acc = [0.62, 0.71, 0.78, 0.81]
new_acc = [0.68, 0.74, 0.80, 0.83]
```

```python
acc_curve = new_acc
plt.plot(acc_curve, marker='o')
plt.title('Accuracy over epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.savefig('accuracy_plot.png')
plt.show()
```

## Results
Baseline reaches 81% and the new algorithm reaches 83% on the validation set.
